**Linear regression** is a *supervised learning algorithm* that models a **dependent variable** \(y\) as a function of one or more **independent variables** (also called *features*) \(x_i\).

The goal of linear regression is to find a **line** (in one dimension) or a **hyperplane** (in higher dimensions) that best *fits* the observed data.

---

### Problem Setup

- **Target variable**:  
  - \( y \) — a numerical value we want to predict

- **Features**:  
  - \( x_1, x_2, ..., x_n \) — input variables describing the object 

#### The main idea of regression is to **find the optimal model coefficients (weights) so that further we can predict unknown target values from known features**.

This is usually done by **minimizing a loss function**, such as the **Mean Squared Error (MSE)**

---

### Intuition

Linear regression assumes a **linear relationship** between the target and the features:

> Small changes in the input features lead to proportional changes in the output.

---

### Examples

- predicting the price of a house using the number of rooms in that house (y: price, x_1: number of rooms) 

OR 
- predicting weight from height and age (y: weight, x_1: height, x_2: age).

### Least Squares Method in Linear Regression

Linear regression is essentially a **Least Squares method**.

The method finds the model coefficients by **minimizing the sum of squared differences** between the actual target values and the predicted values from the linear model.  

These differences are called **residuals**:




By minimizing this sum, the model finds the line (or hyperplane) that **best fits the data** in terms of mean squared error.


## Loss Function

In **Linear Regression**, the **loss function** measures how far the predicted values are from the actual values.  
The smaller the loss, the better the model predicts.

The most common loss function is **Mean Squared Error (MSE)**, which calculates the **average of squared differences** between the actual and predicted values:



The main goal of linear regression is to **find the minimum of the loss function**.  

- Ideally, the loss would be `0`, meaning predictions perfectly match the actual values.  
- In real-world scenarios, this is almost impossible due to noise and variability in the data.  
- Therefore, we aim to **minimize the loss as much as possible**:


## Gradient Descent Method

To minimize the loss function, we can use the **Gradient Descent Method**.

**Gradient Descent** is an optimization technique used to train a linear regression model by **iteratively reducing the prediction error**.

**How it works:**
1. Start with random initial values for the model coefficients (weights).  
2. Calculate the **gradient** of the loss function with respect to each coefficient.  
3. Update the coefficients in the direction that **reduces the loss**:



### What is the difference between Least Squares vs Gradient Descent in Linear Regression?

In linear regression, **Least Squares** and **Gradient Descent** are closely related, but they play **different roles** in the learning process.

---

### Core Difference

They are not competing methods, but **complementary tools**.

> **Least Squares defines the optimization problem,  
> Gradient Descent provides a way to solve it.**

> **In linear regression, Least Squares specifies the criterion for the best model, while Gradient Descent is one of the possible ways to find that model.**

---

### Summary Table

| Aspect | Least Squares | Gradient Descent |
|------|---------------|------------------|
| Role | Defines the objective | Solves the objective |
| Type | Error minimization method | Optimization algorithm |
| Solution | Analytic | Iterative |
| Scalability | Limited for large data | Scales to large datasets |
| Question | *what does “best fit” mean?* | *how do we reach the optimum?* |

## Python Implementation of Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Generating Random Dataset
np.random.seed(42)

X = np.random.rand(50, 1) * 100  

y = 3.5 * X + np.random.randn(50, 1) * 20
```

```python
# Creating and Training Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# Predicting Y Values
y_pred = model.predict(X)
```

```python
# Visualizing the Regression Line
plt.figure(figsize=(6,4)) 
plt.scatter(X, y, color='blue', label='Data Points') 
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line') 
plt.title('Linear Regression on Random Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])
```

If our trained linear regression model has the following parameters:  

- **Slope (coefficient)**: 3.455  
- **Intercept**: 1.934  

Then the model's function can be written as a **linear equation**:

`y = f(x) = 3.4553x + 1.9338`


During inference, to get a prediction from the model, we simply **plug the known feature value** `x` into this formula and compute the result `y`.
