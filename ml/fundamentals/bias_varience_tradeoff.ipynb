8. Что такое bias-variance tradeoff?
Bias-variance tradeoff это фундаментальная концепция в машинном обучении, описывающая баланс между двумя типами ошибок модели. Bias или смещение это ошибка из-за слишком упрощённых предположений модели, когда она не может уловить истинную зависимость в данных. Представьте, что вы пытаетесь описать параболу прямой линией - у вас будет систематическая ошибка, потому что линейная модель слишком проста. Я столкнулся с высоким bias, когда пытался предсказать нелинейную зависимость между температурой и продажами мороженого с помощью простой линейной регрессии. Variance или разброс это ошибка из-за чрезмерной чувствительности модели к обучающим данным, когда она улавливает случайный шум. Модель с высоким variance сильно меняет свои предсказания при небольших изменениях в обучающих данных. Это как студент, который запомнил все конкретные примеры из учебника, но не понял общий принцип. Идеальная модель должна иметь низкий bias и низкий variance, но на практике уменьшение одного часто увеличивает другое. Простые модели типа линейной регрессии имеют высокий bias но низкий variance, они стабильны но могут быть слишком простыми. Сложные модели типа глубоких нейронных сетей имеют низкий bias но высокий variance, они гибкие но могут переобучаться. Регуляризация помогает найти баланс, добавляя штраф за сложность модели. Ensemble методы типа bagging снижают variance, усредняя предсказания нескольких моделей, а boosting снижает bias, последовательно исправляя ошибки. Понимание этого tradeoff помогает мне выбирать правильную сложность модели для конкретной задачи.
