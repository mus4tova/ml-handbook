## The Confusion Matrix Structure

**True Positive (TP)** - model correctly predicted as positive and it is positive in reality *(identifying a spam email as spam)*

**True Negative (TN)** - model correctly predicted as negative and it is negative in reality *(identifying a regular email as not spam)*

**False Positive (FP)** - model incorrectly predicted as positive but it is negative in reality, overprediction *(identifying a regular email as spam)*

**False Negative (FN)** - model incorrectly predicted as negative but it is positive in reality, underprediction *(identifying a spam as a regular email)*

#### Since the main goal of binary classification is to identify the positive class, the error is measured based on predictions labeled as positive.

#### Here, P means **predicted as positive** -> TP indicates that the positive prediction was correct, while FP indicates that the positive prediction was incorrect

---

## Confusion matrix practical example

Let’s create a hypothetical dataset where spam is Positive and not spam is Negative. We have the following data:

Total number of test set: 200 emails

**80 of them are spam**: 

Among 80 spam emails model marked 60 as a spam (TP), 20 as a not spam (FN) - underprediction

**120 are not spam**:

Among 120 non-spam emails model correctly marked as not spam 90 of them (TN), and 30 incorrectly marked as a spam (FP) - overprediction

True Positive: 60
True Negative: 100
False Positive: 30
False Negative: 20

## Accuracy



Accuracy is a tricky metric. 

#### Don't use it when dataset is imbalanced

Example. Detect fraudulent transactions:

- Total transactions: **10,000**
- Normal transactions: **9,900**
- Fraudulent transactions: **100** (1%)

#### A “Bad” Model
The model **always predicts “not fraud.”**

- TP = 0  
- FN = 100  
- TN = 9,900  
- FP = 0  

-> Accuracy = (TP + TN) / Total = 9,900 / 10,000 = 99%


#### Other Scenarios Where Accuracy Can Be Misleading
- 🩺 Rare disease detection  
- 🚨 Intrusion or threat detection  
- 📦 Defect detection in manufacturing  
- 📢 Click-through rate (CTR) or conversion prediction

## Precision



Evaluate prediction class

#### precision = TP / predicted as positive

Compares correctly predicted with the whole prediction class,

shows how many of the truly positive cases model managed to find among all predicted as positives


ex. we predcited 30 spam emails as positives, 20 of them are actually spams (TP), 10 of them are regular emails in reality (FP) -> model correctly predicted in 20/30=67% of case

---



## Recall



Compares prediction with reality class

#### recall = TP / positive in reality

How good we covered actual positive class, how many of actually positives we catch with our model

ex. we have 20 spam emails in reality, model catch 15 of them (TP), 5 of them model didnt managed to catch (FN) -> model 'covered' 15/20=75% of spam emails

## Prevision vs Recall
### When Precision Is More Important

**Precision matters more when false positives are costly.**  
That is, when making an error like *“the model said YES, but it’s actually NO”* causes serious problems.

👉 Key question:  
**“If the model predicts positive, can we trust it?”**

### Examples
- 📧 **Spam filtering** — it’s better to let some spam through than to delete an important email  
- 🧑‍💼 **User blocking** — you must not block legitimate users  
- 📢 **Advertising** — ads should be shown only to users who are very likely to be interested  

🔹 Here, **fewer but more accurate positives** are preferred.

---

## When Recall Is More Important

**Recall matters more when missing a positive case is dangerous.**  
That is, when an error like *“the model said NO, but it was actually YES”* is very costly.

👉 Key question:  
**“How many of the real positive cases did we find?”**

### Examples
- 🩺 **Disease detection** — it’s better to raise a false alarm than to miss a disease  
- 💳 **Fraud detection** — it’s better to review extra transactions than to miss fraud  
- 🚨 **Security systems** — a false alarm is better than a missed threat  
- 🔍 **Defect detection** — defective items must not be missed  

🔹 Here, **finding everything is more important than being perfectly accurate**.
